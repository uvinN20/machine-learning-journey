```python
from sklearn.linear_model import LinearRegression


X = [[1], [2], [3], [4]]


y = [5, 8, 11, 14]


model = LinearRegression()


model.fit(X, y)


print("Weight:", round(model.coef_[0], 2))
print("Bias:", round(model.intercept_, 2))
print("Prediction:", model.predict([[5]]))
```
