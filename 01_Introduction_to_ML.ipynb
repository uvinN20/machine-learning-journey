```python

from sklearn.tree import DecisionTreeClassifier

# Training data: hours studied
X = [[1], [2], [3], [4], [5], [6]]

# 0 = Fail, 1 = Pass
y = [0, 0, 0, 1, 1, 1]

model = DecisionTreeClassifier()
model.fit(X, y)

print("Prediction:", model.predict([[5]]))
```

```python

from sklearn.tree import DecisionTreeClassifier

# X: 1 = contains "FREE"; 0 = does not
X = [[1]] * 10 + [[0]] * 10

# y: 1 = spam; 0 = normal
y = [1]*8 + [0]*2 + [1]*2 + [0]*8

# Create and train the ML model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# Predict two new emails
print(model.predict([[1]]))
print(model.predict([[0]]))
```
