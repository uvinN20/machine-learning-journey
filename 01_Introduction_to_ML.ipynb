```python

from sklearn.tree import DecisionTreeClassifier

# Training data: hours studied
X = [[1], [2], [3], [4], [5], [6]]

# 0 = Fail, 1 = Pass
y = [0, 0, 0, 1, 1, 1]

model = DecisionTreeClassifier()
model.fit(X, y)

print("Prediction:", model.predict([[5]]))
```

```python

from sklearn.tree import DecisionTreeClassifier

# X: 1 = contains "FREE"; 0 = does not
X = [[1]] * 10 + [[0]] * 10

# y: 1 = spam; 0 = normal
y = [1]*8 + [0]*2 + [1]*2 + [0]*8

# Create and train the ML model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# Predict two new emails
print(model.predict([[1]]))
print(model.predict([[0]]))
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.cluster import KMeans


X = [[1], [2], [3], [4]]
y = [0, 0, 1, 1]  # 0 = Fail, 1 = Pass
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)
print("Supervised prediction:", model.predict([[5]]))


data = [[1], [2], [8], [9]]
clusters = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters.fit(data)
centers = sorted(round(float(c[0]), 1) for c in clusters.cluster_centers_)
print("Cluster centers:", centers)
```
