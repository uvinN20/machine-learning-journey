```python

from sklearn.tree import DecisionTreeClassifier

# Training data: hours studied
X = [[1], [2], [3], [4], [5], [6]]

# 0 = Fail, 1 = Pass
y = [0, 0, 0, 1, 1, 1]

model = DecisionTreeClassifier()
model.fit(X, y)

print("Prediction:", model.predict([[5]]))
```
